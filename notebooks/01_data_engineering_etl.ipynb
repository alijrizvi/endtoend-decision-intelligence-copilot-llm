### Initial Data Cleaning & Feature Engineering in Python:

```python
# Revenue Earned by Business per Transaction
df['Revenue'] = (df['UnitPrice']*df['Quantity']).round(2)

# Grouping by Date
# df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

df = df.set_index(df['InvoiceDate'].dt.date) # Setting the Date as the Index
df['DayOfWeek'] = pd.to_datetime(df.index).dayofweek + 1 # Day
df['Month'] = pd.to_datetime(df.index).month # Month
df['RevenueLag_1Day'] = df['Revenue'].shift(1) # Shifts to a Calculation 1 Day Later
df.drop(columns = ['InvoiceDate'], inplace = True) # Removing the Now-Unnecessary Date Column

# Viewing Now
df.head(4)
```

## Starting on the Cleaned **ETL Pipeline** in SQL:

Using Conditional Logic and Segmentation, Structured Data Manipulation, and Advanced Window Functions among other techniques - including and not limited to `CASE WHEN`s, Complex `JOIN`s, and `CTE`s and `Temp Tables` - we will create a standardized pipeline for Data Manipulation and Storage for Subsequent Analyses and Engineering of the Ultimate Copilot tool.

```python
import time

# Cleaning Up the Table & then Viewing
print("Creating/Replacing 'clean_ecomm' table with all transformations...")

start_time = time.time() # Initial time at starting the Process

con.execute(""" CREATE OR REPLACE TABLE clean_ecomm AS
             SELECT
                REGEXP_REPLACE(CAST(InvoiceNo AS VARCHAR), '[^a-zA-Z0-9]', '', 'g') AS InvoiceNo,
                REGEXP_REPLACE(CAST("CustomerID" AS VARCHAR), '[^a-zA-Z0-9]', '', 'g') AS CustomerID,
                REGEXP_REPLACE(CAST(StockCode AS VARCHAR), '\\D', '', 'g') AS StockCode,
                CASE
                  WHEN TRIM(UPPER(Country)) LIKE 'UNITED K%' THEN 'United Kingdom'
                  ELSE NULL
                END AS Country, -- Ensuring that only the UK's Values Appear after Filtering
                -- Cleaning Description: Removing non-Alphanumeric characters, handling empty/short Descriptions
                CASE
                  WHEN Description IS NULL OR LENGTH(TRIM(Description)) = 0 OR LENGTH(TRIM(Description)) < 3 THEN NULL -- Setting a minimum length to avoid garbage descriptions
                  ELSE REGEXP_REPLACE(Description, '[^a-zA-Z0-9 ]', '', 'g')
                END AS Description,
                CAST(Quantity AS NUMERIC(12, 2)) AS Quantity,
                CAST(UnitPrice AS NUMERIC(10, 2)) AS Price,
                CAST(Revenue AS NUMERIC (10, 2)) AS Revenue,
                CAST(DayOfWeek AS STRING) AS DayOfWeek,
                CAST(Month AS INTEGER) AS Month,
                CAST(RevenueLag_1Day AS NUMERIC(10, 2)) AS RevenueLag_1Day,
                -- Feature Engineering PurchaseID
                CONCAT(
                  LEFT(CAST("CustomerID" AS VARCHAR), COALESCE(STRPOS(CAST("CustomerID" AS VARCHAR), '.'), LENGTH(CAST("CustomerID" AS VARCHAR)) + 1) - 1),
                  '-',
                  REGEXP_REPLACE(CAST(StockCode AS VARCHAR), '\\D', '', 'g')
                ) AS PurchaseID,
                -- Feature Engineering UnitPrice, handling division by zero
                CASE
                  WHEN Quantity > 0 THEN ROUND(UnitPrice / Quantity, 2)
                  ELSE 0.00 -- Assign 0 or NULL if Quantity is zero or negative
                END AS UnitPrice
              FROM ecomm; """
            )

end_time = time.time() # Time after the whole ETL Process was Complete
elapsed_time = end_time - start_time

print(f"Time taken to Create the Table: {elapsed_time:.2f} seconds.")

# It is Crucial for Businesses to note the Time taken in the ETL Process, as this helps Optimize the Stage further and
# getting all Data in as swiftly and supremely, quality-wise, ready for analysis, is Critical
```

### Further ETL & Data Organization and Standardization with Decision Logic:

```python
# Common Table Expressions (CTE) + Subquery: [Customer Segmentation] Creating a Distinct Table containing the most Profitable Customers Ranked by Expenditure from a Pool who Spent, in Total, Above Average

con.execute(""" CREATE OR REPLACE TABLE above_avg_customers AS
WITH CustomerSpend AS (
    SELECT
        CustomerID,
        SUM(Revenue) AS TotalSpent
    FROM clean_ecomm
    WHERE CustomerID IS NOT NULL -- Exclude transactions without a customer ID from spend calculation
    GROUP BY CustomerID
),
AverageOverallSpend AS (
    SELECT AVG(TotalSpent) AS AvgSpend
    FROM CustomerSpend
)
SELECT
    cs.CustomerID,
    cs.TotalSpent,
    DENSE_RANK() OVER (ORDER BY cs.TotalSpent DESC) AS Ranking
FROM CustomerSpend cs
WHERE cs.TotalSpent > (SELECT AvgSpend FROM AverageOverallSpend); """).fetchdf()
```

```python
# Always Confirming: Yes, the Table looks good
con.execute(""" SELECT *
             FROM above_avg_customers LIMIT 7 """).fetchdf()
```

```python
# Creating a Distinct Table containing Performance Metrics of each Product (Identified by "StockCode")

start_time = time.time()

con.execute(""" CREATE OR REPLACE TABLE product_stats AS
                SELECT
                  StockCode,
                  Month,
                  COUNT(*) as Transactions,
                  CAST(SUM(Quantity) AS INTEGER) AS TotalSold,
                  SUM(Revenue) as TotalRevenue
                FROM clean_ecomm
                WHERE StockCode IS NOT NULL AND StockCode != ''
                GROUP BY StockCode, Month
                ORDER BY StockCode, Month
                LIMIT 10; """).fetchdf()

end_time = time.time()

elapsed_time = end_time - start_time

print(f"Time taken to Create the Product Stats Table: {elapsed_time:.2f} seconds.")
```

```python
# Check

con.execute("""
            SELECT *
            FROM product_stats
            LIMIT 7; """).fetchdf()
```

```python
# Advanced/Complex JOIN: Enriching Transactions with Monthly Product Performance
# Creating a new, "Final" Table by Joining the Original "clean_ecomm" with its Monthly Average Revenue per Product


start_time = time.time()

con.execute(""" CREATE OR REPLACE TABLE final_ecomm AS
    WITH MonthlyProductAvg AS (
      SELECT
          StockCode,
          Month,
          ROUND(AVG(Revenue), 2) AS AvgMonthlyProductRevenue,
          ROUND(AVG(UnitPrice), 2) AS AvgMonthlyUnitPrice,
          CAST(AVG(Quantity) AS INTEGER) AS AvgMonthlyQuantitySold
      FROM clean_ecomm
      WHERE StockCode IS NOT NULL AND Revenue IS NOT NULL AND Month IS NOT NULL
      GROUP BY StockCode, Month
)
    SELECT
        ce.*, -- Selecting all Columns from clean_ecomm
        mpa.AvgMonthlyProductRevenue,
        mpa.AvgMonthlyUnitPrice,
        mpa.AvgMonthlyQuantitySold
    FROM clean_ecomm ce
    LEFT JOIN MonthlyProductAvg mpa
        ON ce.StockCode = mpa.StockCode AND ce.Month = mpa.Month; """)

end_time = time.time()

# Of course, Noting the Time taken for this Process too - Efficient, Optimized Systems are key!
elapsed_time = end_time - start_time

print(f"Time Taken to Create the Final, Aggregated Table: {elapsed_time:.2f} seconds.")
```
